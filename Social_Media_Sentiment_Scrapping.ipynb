```python
from bs4 import BeautifulSoup as bs
import requests
import pandas as pd
```

```python
with open("/content/drive/MyDrive/Messy_Sentiment_Data.html", 'r',encoding = "utf-8") as f:
  html_ = bs(f,'html.parser')
print(html_)
print(html_.prettify)
```

```python
headers_list = []
th = html_.find_all('th')
for i in th:
  headers_list.append(i.text)
print(headers_list)
```

```python
data_list = []
tr = html_.find_all('tr')
for i in tr:
  td = i.find_all('td')
  row_data = [tds.get_text(strip = True) for tds in td]
  if row_data:
    data_list.append(row_data)
print(data_list)
```

```python
Client_Progress_dataset = [dict(zip(headers_list,row)) for row in data_list]
print(Client_Progress_dataset)
```

```python
client_Progress_dataset = pd.DataFrame(Client_Progress_dataset)
```

```python
client_Progress_dataset.to_csv('Client_Progress_dataset_Scrap.csv')
```
